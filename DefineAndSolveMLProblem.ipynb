```python
# YOUR CODE HERE


# 'host_is_superhost' is included in the features list for convenience
# so I can create a copy of the data frame with the features I need and it is store in df_selected .
# I will drop it later to form X (features) and use it as y (label).
features = ['host_response_rate', 'host_listings_count', 'host_has_profile_pic', 'host_identity_verified', 'neighbourhood_group_cleansed', 'room_type', 'reviews_per_month', 'review_scores_rating', 'review_scores_cleanliness', 'review_scores_checkin', 'review_scores_communication', 'review_scores_value','has_availability','instant_bookable']
df_selected = df[features].copy()
df_selected.isnull().sum()

#that line helps us take a look at some missing value within the data frame 
df_selected.loc[df_selected['host_response_rate'].isnull()]

df_selected['host_response_rate'].describe()

# After checking the mean, median, and distribution of 'host_response_rate',
# we saw that the data is skewed to the left (mean < median), which means the outliers are affecting the mean.
# To avoid those outliers affecting the fill value, we replace missing values with the median.
median_host_response_rate = df_selected['host_response_rate'].median()
df_selected['host_response_rate'].fillna(value=median_host_response_rate, inplace=True)

df_selected.isnull().sum()



# Pick the columns that have text values (object type)
to_encode_obj = df_selected.select_dtypes(include='object')
to_encode = list(to_encode_obj.columns) 
print(to_encode) 

# Check how many different values each column has
unique_value = df_selected[to_encode].nunique()
print(unique_value) 

# Import OneHotEncoder to turn text columns into numbers
from sklearn.preprocessing import OneHotEncoder
enc = OneHotEncoder(sparse_output=False, handle_unknown='ignore')  

# Fit and transform the text columns into numbers
encoded = enc.fit_transform(df_selected[to_encode])  

# Get the new column names that were created from one-hot encoding
feature_names = enc.get_feature_names_out(to_encode)

# Turn the encoded values into a new DataFrame with proper column names
df_encoded = pd.DataFrame(encoded, columns=feature_names, index=df_selected.index)

#  Remove the old text columns from our data
df_selected = df_selected.drop(columns=to_encode)

# Add the new encoded columns back to our data
df_selected = pd.concat([df_selected, df_encoded], axis=1)


df_selected.head(10)

# checking for imbalance 


# When I checked the imbalance for the three columns below, I saw they only contain True values.
# Since there's no variation, they won't help the model learn anything useful.
# So I decided to remove these features from the data.
df['host_has_profile_pic'].value_counts()
df['host_identity_verified'].value_counts()
df['has_availability'].value_counts()
drop_columns = ['host_has_profile_pic', 'host_identity_verified', 'has_availability']
df_selected = df_selected.drop(columns=drop_columns)
df_selected.dtypes

df_selected['instant_bookable'].value_counts()




```

```python
# YOUR CODE HERE
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (accuracy_score, log_loss, roc_curve, auc, confusion_matrix)
from sklearn.feature_selection import SelectKBest, f_classif
import matplotlib.pyplot as plt
import seaborn as sns
```

<b>Task:</b> Use the rest of this notebook to carry out your project plan. 

You will:

1. Prepare your data for your model.
2. Fit your model to the training data and evaluate your model.
3. Improve your model's performance by performing model selection and/or feature selection techniques to find best model for your problem.

Add code cells below and populate the notebook with commentary, code, analyses, results, and figures as you see fit. 

```python
# YOUR CODE HERE



# First, we split the data into X for the features and y for the label
X = df_selected.drop(columns='instant_bookable')
y = df_selected['instant_bookable']
# Split the data into training and testing sets 75% train and 25% test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# # Create a logistic regression model
model_LR = LogisticRegression(max_iter=1000)
model_LR.fit(X_train, y_train)
LR_probability_predictions = model_LR.predict_proba(X_test)
LR_class_predictions = model_LR.predict(X_test)
loss_default = log_loss(y_test, LR_probability_predictions)
acc_LR = accuracy_score(y_test, LR_class_predictions)
fpr_default, tpr_default, _ = roc_curve(y_test, LR_probability_predictions[:, 1])
auc_default = auc(fpr_default, tpr_default)
print("Log loss (Default Logistic Regression):", loss_default)
print("Accuracy (Default Logistic Regression):", acc_LR)
print("AUC (Default Logistic Regression):", auc_default, "\n")

# Tune in to find the best hyperparameter for C
param_grid_LR = {'C': [0.01, 0.1, 1, 10, 100]}
grid_LR = GridSearchCV(LogisticRegression(max_iter=1000), param_grid= param_grid_LR, cv=5, scoring='accuracy')
grid_LR.fit(X_train, y_train)
model_best = grid_LR.best_estimator_
LR_best_C = grid_LR.best_params_['C']
class_label_predictions_best = model_best.predict(X_test)
LR_best_probability_prediction = model_best.predict_proba(X_test)
fpr_best, tpr_best, _ = roc_curve(y_test, LR_best_probability_prediction[:, 1])
auc_best = auc(fpr_best, tpr_best)

# Feature Selection
selector = SelectKBest(score_func=f_classif, k=5)
selector.fit(X, y)
top_features = X.columns[selector.get_support()]
X_train_top = X_train[top_features]
X_test_top = X_test[top_features]

# Logistic Regression - Best on Top 5 Features
print("Retraining Best Logistic Regression on Top 5 Features")
model_top_best = LogisticRegression(C=LR_best_C, max_iter=1000)
model_top_best.fit(X_train_top, y_train)
class_label_predictions_top_best = model_top_best.predict(X_test_top)
prob_top_best = model_top_best.predict_proba(X_test_top)
loss_top_best = log_loss(y_test, prob_top_best)
acc_top_best = accuracy_score(y_test, class_label_predictions_top_best)
print("Log loss (Best Logistic on Top 5 Features):", loss_top_best)
print("Accuracy (Best Logistic on Top 5 Features):", acc_top_best)
proba_top_best = prob_top_best[:, 1]
fpr_top_best, tpr_top_best, _ = roc_curve(y_test, proba_top_best)
auc_top_best = auc(fpr_top_best, tpr_top_best)
print("AUC (Best Logistic on Top 5 Features):", auc_top_best, "\n")

# Decision Tree - Default
print("Training Decision Tree Classifier (default)...")
model_DT = DecisionTreeClassifier()
model_DT.fit(X_train, y_train)
class_label_predictions_DT = model_DT.predict(X_test)
acc_DT = accuracy_score(y_test, class_label_predictions_DT)
print("Accuracy (Default Decision Tree):", acc_DT)
DT_proba = model_DT.predict_proba(X_test)
fpr_dt_default, tpr_dt_default, _ = roc_curve(y_test, DT_proba[:, 1])
auc_dt_default = auc(fpr_dt_default, tpr_dt_default)

# Decision Tree - GridSearchCV
print("Performing GridSearchCV for Decision Tree...")
param_grid_dt = {'max_depth': [4, 8, 16], 'min_samples_leaf': [1, 5, 10]}
grid_search_dt = GridSearchCV(
       estimator=DecisionTreeClassifier(),  
       param_grid=param_grid_dt,
       cv=5,
       scoring='accuracy',
       refit=True
)
grid_search_dt.fit(X_train, y_train)
dt_best = grid_search_dt.best_estimator_
class_label_predictions_DT_best = dt_best.predict(X_test)
acc_DT_best = accuracy_score(y_test, class_label_predictions_DT_best)
print("Best params (Decision Tree):", grid_search_dt.best_params_)
print("Accuracy (Best Decision Tree):", acc_DT_best)
DT_best_proba = dt_best.predict_proba(X_test)
fpr_dt_best, tpr_dt_best, _ = roc_curve(y_test, DT_best_proba[:, 1])
auc_dt_best = auc(fpr_dt_best, tpr_dt_best)

# Decision Tree - Best on Top 5 Features
print("Retraining Best Decision Tree on Top 5 Features...")
dt_top_best = DecisionTreeClassifier(max_depth=dt_best.max_depth, min_samples_leaf=dt_best.min_samples_leaf)
dt_top_best.fit(X_train_top, y_train)
class_label_predictions_DT_top = dt_top_best.predict(X_test_top)
acc_DT_top = accuracy_score(y_test, class_label_predictions_DT_top)
print("Accuracy (Best DT on Top 5 Features):", acc_DT_top)
DT_top_proba = dt_top_best.predict_proba(X_test_top)
fpr_dt_top, tpr_dt_top, _ = roc_curve(y_test, DT_top_proba[:, 1])
auc_dt_top = auc(fpr_dt_top, tpr_dt_top)

# Plot ROC Curves using seaborn for all models
# Logistic Regression
sns.lineplot(x=fpr_default, y=tpr_default, color='green', label='Default Logistic')
sns.lineplot(x=fpr_best, y=tpr_best, color='red', label='Tuned Logistic')
sns.lineplot(x=fpr_top_best, y=tpr_top_best, color='blue', label='Tuned + Top 5 Features Logistic')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve - Logistic Regression Models')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Decision Tree
sns.lineplot(x=fpr_dt_default, y=tpr_dt_default, color='purple', label='Default Decision Tree')
sns.lineplot(x=fpr_dt_best, y=tpr_dt_best, color='orange', label='Tuned Decision Tree')
sns.lineplot(x=fpr_dt_top, y=tpr_dt_top, color='cyan', label='Tuned + Top 5 Features DT')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve - Decision Tree Models')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


```
